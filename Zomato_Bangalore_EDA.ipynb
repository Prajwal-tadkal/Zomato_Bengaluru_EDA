```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv("zomato.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop(['url','phone'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df['rate'].unique()
```

```python
df['rate'] = df['rate'].replace(
    ['NEW','-'],
    np.nan
)
```

```python
df['rate'] = df['rate'].str.split('/').str[0]
```

```python
df['rate'].head()
```

```python
df['rate'] = df['rate'].astype(float)
```

```python
df['rate'].dtype
```

```python
df['approx_cost(for two people)'].head()
```

```python
df['approx_cost(for two people)'] = \
df['approx_cost(for two people)'].astype(str)

df['approx_cost(for two people)'] = \
df['approx_cost(for two people)'].str.replace(',','')
```

```python
df['approx_cost(for two people)'] = \
df['approx_cost(for two people)'].astype(float)
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['rate'],
    bins=20,
    kde=True
)

plt.title("Restaurant Ratings Distribution")

plt.show()
```

```python
sns.countplot(
    x='online_order',
    data=df
)

plt.title("Online Ordering Availability")

plt.show()
```

```python
sns.countplot(
    x='book_table',
    data=df
)

plt.title("Table Booking Availability")

plt.show()
```

```python
top_locations = df['location'].value_counts().head(10)

top_locations
```

```python
plt.figure(figsize=(12,6))

top_locations.plot(kind='bar')

plt.title("Top 10 Restaurant Locations")

plt.show()
```

```python
cuisine = (
    df['cuisines']
    .dropna()
    .str.split(',')
    .explode()
)
```

```python
cuisine.value_counts().head(10)
```

```python
cuisine.value_counts().head(10).plot(
    kind='bar'
)

plt.title("Top Cuisines")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['approx_cost(for two people)'],
    bins=40
)

plt.title("Cost Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='approx_cost(for two people)',
    y='rate',
    data=df
)

plt.show()
```
